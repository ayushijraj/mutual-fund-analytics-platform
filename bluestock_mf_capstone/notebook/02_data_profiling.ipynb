```python
import pandas as pd
import os

os.chdir(r"C:\Users\hp\bluestock_mf_capstone\data\raw")

print(os.getcwd())
```

```python
fund_master = pd.read_csv("01_fund_master.csv")
nav_history = pd.read_csv("02_nav_history.csv")
scheme_perf = pd.read_csv("07_scheme_performance.csv")
transactions = pd.read_csv("08_investor_transactions.csv")
```

```python
# Check Data types
for name, df in {
    "fund_master": fund_master,
    "nav_history": nav_history,
    "scheme_performance": scheme_perf,
    "transactions": transactions
}.items():
    
    print("\n" + "="*50)
    print(name.upper())
    print(df.info())
```

```python
# Check Missing Values
for name, df in {
    "fund_master": fund_master,
    "nav_history": nav_history,
    "scheme_performance": scheme_perf,
    "transactions": transactions
}.items():
    
    print("\n" + "="*50)
    print(name.upper())
    print(df.isnull().sum())
```

```python
# Check Duplicate Records
for name, df in {
    "fund_master": fund_master,
    "nav_history": nav_history,
    "scheme_performance": scheme_perf,
    "transactions": transactions
}.items():
    
    print("\n" + "="*50)
    print(name.upper())
    print("Duplicates:", df.duplicated().sum())
```
